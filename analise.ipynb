```python
import pandas as pd

tabela = pd.read_csv("gastos.csv")
display(tabela)
```

```python
display(tabela.groupby("categoria")["valor"].sum().sort_values(ascending=False))
```

```python
total_gasto = tabela["valor"].sum()
media_transacao = tabela["valor"].mean()

display(pd.Series({
    "total_gasto_geral": total_gasto,
    "gasto_medio_por_transacao": media_transacao
}))
```

```python
import matplotlib.pyplot as plt

total_por_categoria = tabela.groupby("categoria")["valor"].sum().sort_values(ascending=False)
total_por_categoria.plot(kind="bar")
```

```python
import matplotlib.pyplot as plt

total_por_categoria = tabela.groupby("categoria")["valor"].sum().sort_values(ascending=False)
total_por_categoria.plot(kind="pie")
```

```python
total_gasto = tabela["valor"].sum()
media_transacao = tabela["valor"].mean()

pd.Series({
    "total_gasto_geral": total_gasto,
    "gasto_medio_por_transacao": media_transacao
})

total_por_categoria = tabela.groupby("categoria")["valor"].sum().sort_values(ascending=False)

print(f"Total gasto: R${total_gasto:.2f}")
print(f"Maior gasto: {total_por_categoria.index[0]}")
print(f"Menor gasto: {total_por_categoria.index[-1]}")
print(f"Média por transação: R$ {media_transacao:.2f}")
```
